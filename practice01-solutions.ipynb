### Задача 3

Напишите код, который псевдослучайным образом выбирает значения `b0` и `b1`, строит график с соответствующей прямой, показывает его пользователю и запрашивает с клавиатуры значения `b0` и `b1`. Если ответы верные, то выводится сообщение `Молодец!`, если нет – `Ошибка в b0` или `Ошибка в b1`, в зависимости от того, где ошибся пользователь.

```python
# копируем код из предыдущих задач для выбора b0, b1
# и отрисовки графика

numbers = list(range(1, 6)) + list(range(-5, 0))
b0 = np.random.choice(numbers)
b1 = np.random.choice(numbers)
x = np.arange(-5, 6)
y = b0 + b1 * x
fig, ax = plt.subplots()
ax.set_xlim((-5, 6));
ax.set_ylim((-5, 6));
ax.set_xticks(range(-5, 6));
ax.set_yticks(range(-5, 6));
ax.grid();
ax.vlines(0, ymin = -5, ymax = 6, linewidth = 0.5, color = "k");
ax.hlines(0, xmin = -5, xmax = 6, linewidth = 0.5, color = "k");
ax.plot(x, y);

# добавляем строку для вывода графика на экран

plt.show()

# пишем код для запроса коэффициентов с клавиатуры и их проверки

answer1 = int(input())
answer2 = int(input())
if answer1 == b0 and answer2 == b1:
    print("Молодец!")
if answer1 != b0:
    print("Ошибка в b0, верное значение", b0)
if answer2 != b1:
    print("Ошибка в b1, верное значение", b1)
```

**Дополнение.** Тренажер можно организовать по-другому:

* в цикле создаем 5 разных графиков и сохраняем их в файлы `.png`;
* одновременно верные значения `b0` и `b1` сохраняем в список `answers`;
* открываем файлы `1.png`-`5.png`, себе на листок записываем уравнения прямых;
* сверяем ответы со списком `answers`.

```python
# создаем список answers для верных b0, b1
answers = []

for i in range(1, 6):
    
    # весь код для выбора чисел и построения прямой
    numbers = list(range(1, 6)) + list(range(-5, 0))
    b0 = np.random.choice(numbers)
    b1 = np.random.choice(numbers)
    x = np.arange(-5, 6)
    y = b0 + b1 * x
    fig, ax = plt.subplots()
    ax.set_xlim((-5, 6));
    ax.set_ylim((-5, 6));
    ax.set_xticks(range(-5, 6));
    ax.set_yticks(range(-5, 6));
    ax.grid();
    ax.vlines(0, ymin = -5, ymax = 6, linewidth = 0.5, color = "k");
    ax.hlines(0, xmin = -5, xmax = 6, linewidth = 0.5, color = "k");
    ax.plot(x, y);

    # закрываем график, чтобы не показывался здесь
    plt.close();

    # выгружаем в файл с названием вида i.png
    fig.savefig(str(i) + ".png")

    # добавляем в список answers значения
    answers.append([int(b0), int(b1)])

print(answers)
```

Регрессионные прямые добавим вместе, тут весьма специфический код, надо его осмыслить!

> Название функции `polyfit()` говорящее: `poly` – от слова *polynome*, многочлен, `fit` – глагол «подгонять», то есть оценивать модель на имеющихся данных. Когда мы добавляем регрессионную прямую на диаграмму рассеивания, мы подгоняем линейную модель под наши данные (значения `x` и значения `y`), чтобы получить коэффициенты `b1` и `b0` для прямой $y = b_0 + b1_x$. Линейная функция записывается как многочлен первой степени, поэтому в функции `polyfit()` в аргументе `deg` (от *degree*, степень)  мы указываем 1.

```python
b1, b0 = np.polyfit([3, 5, 10], [9, 13, 23], deg = 1)
print(f"y = {b0} + {b1}x")
```

> Проблема: функция `polyfit()`, как и многие функции в `numpy` не умеет игнорировать пропуски в данных. А значит, в случае наличия пропущеных значений `NaN` она не сможет получить коэффициенты `b1` и`b0` как в примере выше. Одна из возможных стратегий – заполнить пропуски средним значением по столбцу. Не идеальная стратегия, здесь логичнее было бы заполнить пропуски средним значением по каждой стране (усреднить за годы) или продублировать значения за ближайшие к пропуску годы, но пока оставим так.

```python
# x1.fillna(x1.mean()) – замени пропуски в x1 на среднее по x1,
# и так для всех

y = y.fillna(y.mean())
x1 = x1.fillna(x1.mean())
x2 = x2.fillna(x2.mean())
x3 = x3.fillna(x3.mean())
x4 = x4.fillna(x4.mean())
```

```python
fig, ax = plt.subplots(figsize = (9, 9), dpi = 300, 
                       nrows = 2, ncols = 2)

# код для диаграмм рассеивания
ax[0][0].scatter(x1, y, color = "w", edgecolor = "k");
ax[0][1].scatter(x2, y, color = "w", edgecolor = "k");
ax[1][0].scatter(x3, y, color = "w", edgecolor = "k");
ax[1][1].scatter(x4, y, color = "w", edgecolor = "k");

# код для добавления прямых
# b –  угол наклона b1, a – константа b0

b1, a1 = np.polyfit(x1, y, deg = 1)
ax[0][0].plot(x1, a1 + b1 * x1, color = "k");

b2, a2 = np.polyfit(x2, y, deg = 1)
ax[0][1].plot(x2, a2 + b2 * x2, color = "k");

b3, a3 = np.polyfit(x3, y, deg = 1)
ax[1][0].plot(x3, a3 + b3 * x3, color = "k");

b4, a4 = np.polyfit(x4, y, deg = 1)
ax[1][1].plot(x4, a4 + b4 * x4, color = "k");
```
