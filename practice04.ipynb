## Теория: случайные величины с библиотекой `scipy`

Импортируем все необходимое:

```python
import numpy as np
from matplotlib import pyplot as plt
from scipy import stats
```

```python
t = stats.t(df = 5)
```

```python
print(t.expect())
print(t.std())
```

```python
print(t.pdf(0))
print(t.pdf([-1, 0, 1]))
```

```python
x = np.linspace(-5, 5, 100)
density = t.pdf(x)

fig, ax = plt.subplots()
ax.plot(x, density);
ax.grid();
```

```python
print(t.cdf(3))
print(1 - t.cdf(3))
print(2 * (1 - t.cdf(3)))
```

```python
print(t.ppf(0.975))
print(t.ppf(0.025))
```
