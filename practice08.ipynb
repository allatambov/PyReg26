### Часть 2: еще немного про эксперименты и t-test

В статье *Pianzola et al. The Impact of Personalized Information on Vote Intention: Evidence from a Randomized Field Experiment* (данные и статья [здесь](https://disk.yandex.ru/d/Am0PKP_tEahnQA)) авторы ставят любопытную задачу: изучить влияние «избирательного компаса» (приложения для сопоставления своих политических взглядов с программой партий) на склонность голосовать за разные партии и изменять свои предпочтения. 

Устройство эксперимента, в котором участвовали студенты четырех швейцарских университетов:

* За месяц до выборов участники заполняют анкету о своих политических предпочтениях и оценивают в шкале от 1 до 10 свою склонность проголосовать за каждую из партий. После выборов участники заполняют такую же анкету, что и за месяц до выборов.
* Участников делят случайным образом на две группы. Экспериментальная группа получает email с приглашением пройти тест на платформе *smartvote*, которая сопоставляет ответы избирателя с программами партий и кандидатов и выдает персонализированный рейтинг совпадений. Контрольная группа такого приглашения не получает.
* Так как не все получившие email с приглашением воспользовались сервисом *smartvote*, далее сравниваются группы тех, кто воспользовался сервисом или нет.

Сравнений было много, мы сфокусируемся на показателе `diff_top_ptv` – разница в оценках склонности проголосовать за самую предпочтительную партию после выборов и до. Если разница положительная и большая – после объявления результатов выборов избиратель утвердился в своем выборе, стал еще больше поддерживать партию, которую поставил до выборов на первое место.

Итого: правда ли, что те, кто использовал сервис `smartvote`, склонны в большей степени поддерживать партию, которую до выборов указали в качестве самой предпочительной?

Проделаем ровно то же самое, но с помощью линейной модели:

```python
m01 = smf.ols("diff_top_ptv ~ smartvote", data = df).fit()
print(m01.summary())
```

**Дополнение.** Авторы сначала строят ровно такую модель, что и выше, а затем справедливо добавляет в нее контрольные переменные – переменные, которые могут влиять как на склонность избирателя изменять свое мнение относительно партии (зависимая переменная `diff_top_ptv`), так и на его решение использовать платформу `smartvote`. В комментариях к таблице 4 (страница 842 в статье) указан перечень таких переменных.

```python
# вот он, список названий столбцов controls
# можно выбрать их из таблицы и оценить масштабы

controls = ['age', 'female', 'highschool', 'bachelor_vocational',
            'master_vocational', 'bachelor_university',
            'master_university', 'phd', 'bern', 'lucerne', 'zurich', 
            'social', 'medicine', 'natural', 'law', 'econ', 
            'pi_not', 'pi_rnot', 'pi_r', 'pi_very']
df[controls].head()
```

Подход, при котором мы записываем уравнение модели в `ols()` в виде формулы, кажется уже не очень удобным. Однако мы можем склеить названия столбцов из `controls` через `+` и собрать строку с уравнением:

```python
equation = "diff_top_ptv ~ smartvote + " + " + ".join(controls)
print(equation)
```

```python
m02 = smf.ols(equation, data = df).fit()
print(m02.summary())
```
