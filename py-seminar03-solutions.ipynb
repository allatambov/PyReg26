## Семинар 3. Описательные статистики в NumPy

### Немного теории: методы и функции для описательных статистик

```python
import numpy as np
```

```python
sample = np.array([20, 40, 60, 40, 45, 30, 
                   25, 25, 30, 40, 45, 50])
```

Методы:

```python
print("Минимум:", sample.min())
print("Максимум:", sample.max())
print("Среднее:", sample.mean())
```

```python
# смещенная оценка дисперсии (с делением на n)

print("Дисперсия:", sample.var())
print("Стандартное отклонение:", sample.std())
```

```python
# несмещенная оценка дисперсии (с делением на n-1)

print("Дисперсия:", sample.var(ddof=1))
print("Стандартное отклонение:", sample.std(ddof=1))
```

Функции:

```python
print(sample.median())
```

> Код выше не сработал, поскольку на массивах, действительно, не определен метод `.median()`. Однако есть функция с таким названием. То же самое с квантилями – только для их вычисления потребуется еще указать уровень в виде числа от 0 до 1. 

```python
print(np.median(sample))
print(np.quantile(sample, 0.25)) # нижний квартиль
print(np.quantile(sample, 0.75)) # верхний квартиль
```

**Утешающий факт.** Если кажется, что в методах и функциях легко запутаться, для вычисления описательных статистик можно использовать функции и всегда писать их перед названием массива:

```python
# всегда np. и название функции

print(np.min(sample))
print(np.max(sample))
print(np.mean(sample))
print(np.var(sample))
print(np.std(sample))
```

**Дополнительно.** В качестве аргумента в `.quantile()` можно указать не один уровень квантиля, а сразу несколько. Но тогда их нужно оформить в виде списка:

```python
# квантиль уровня 0 – минимум, уровня 1 – максимум

print(np.quantile(sample, [0.25, 0.75])) 
print(np.quantile(sample, [0, 0.25, 0.75, 1]))
```

Если нужен перечень уровней с фиксированным шагом, можно воспользоваться функцией `np.arange()`. Например, для получения децилей выставим шаг 0.1:

```python
# старт в 0, финиш в 1.1, шаг 0.1

print(np.arange(0, 1.1, 0.1))
```

Подставляем массив из уровней квантилей и получаем децили:

```python
# нулевой дециль – это просто минимум 20
# десятый дециль – это просто максимум 60
# 10% значений не больше 25
# 20% значений не больше 26
# 30% значений не больше 30, итд

np.quantile(sample, np.arange(0, 1.1, 0.1))
```

* оставьте в массиве `data` только типичные значения.

```python
data = np.array([-0.3, -0.1, 0.5, -0.5, 1., 5.2, 10.6, 1.4, 1.,  
                 0., 0.3, 1.6, 1.4, -0.7, -0.2, 0.3, 0.5, 0.8,  
                 6.2, -2.4, 0.6, -1.9, -2.6, -1.6, -1.6, -0.,
                 -1.3, 0.7, -0., 1.3, 1., 0.1, 0.8, -1., 
                 -0.2, 4.5, 0., -0.7, -0.6, 0.6, 0.7, -0.5, 
                 -1.3, -0.8, 0., 0.4, -0.4, 0.6, 1.1, 0.9, 
                 -1.5, 0.4, -0.6, 1.1, -0.4, 1., -3.8, -0.9, 
                 -1.2, -1.9, -1.5, 1.2, 0.7, -3.6, 1.7, -1.1,
                 -0.3, -1.2, -0., 0.1, 0.6, 0.3, -0.3, -0.3, 
                 -0.4, -0.3, -0.8, 0.2, 0.8, -0.9, 1.1, 2.1, 
                 -1.2, 1.1, 0.3, 0.3, 0.4, -0.1, 0.4, 0.1, 
                 1.5, 0.4, 3.3, -0.3, 0.6, 0.7, -1.9, -1.5, 
                 -6.1, 0.4, 0.9, -0.5, -0.4, 0.1, 1.7, -0.8])
```

```python
Q1 = np.quantile(data, 0.25)
Q3 = np.quantile(data, 0.75)
IR = Q3 - Q1
left = Q1 - 1.5 * IR
right = Q3 + 1.5 * IR
```

```python
# нетипичные значения

outliers = data[(data < left) | (data > right)]
print(outliers)
print(len(outliers))
```

```python
# типичные значения
# выводим в отсортированном виде, 
# проверяем, что нетипичных не осталось

data = data[(data >= left) & (data <= right)]
print(sorted(data)) 
```

### Задача 3

В выборке `income` сохранены значения заработной платы сотрудников. Вычислите децильный коэффициент для оценки неравенства доходов, который получается делением среднего дохода самых 10% богатых на средний доход 10% самых бедных.

```python
income = np.array([ 70000,  90000,  61000,  60000, 150000, 155000, 160000, 162000,
       156000, 130000, 137000,  45000,  48000,  51000,  49500,  53000,
        65000,  63000,  64000,  80000,  82000,  58000])
```

```python
# 10% самых богатых – значения выше 9-го дециля
# 10% самых бедных – значения ниже 1-го дециля

rich = income[income > np.quantile(income, 0.9)]
poor = income[income < np.quantile(income, 0.1)]
print(rich)
print(poor)
```

```python
# считаем средние по rich и poor, затем коэффициент
coef = rich.mean() / poor.mean()
print(coef)
```
