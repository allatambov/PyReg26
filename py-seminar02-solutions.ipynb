### Немного теории: объединение условий

```python
import numpy as np

sample = np.array([5, 10, 7, 45, 32, 12, 19, 17, 19, 10, 24, 30, 3])
```

```python
# оператор И (and)
# одновременное выполнение условий – оба верны
# выбираем элементы более 10 и не более 30

chosen_one = sample[(sample > 10) & (sample <= 30)]
print(chosen_one)
```

```python
# оператор ИЛИ (or)
# верно хотя бы одно условие: либо первое, либо второе, либо оба,
# выбираем четные элементы или кратные 3

chosen_two = sample[(sample % 2 == 0) | (sample % 3 == 0)]
print(chosen_two)
```

> Оператор `%` вычисляет остаток от деления на число. Например, код `56 % 7` вернет число 0, так как 56 делится на 7 без остатка, а код `56 % 3` вернет число 2, так как остаток от деления 56 на 3 равен 2 (на 3 нацело делится 54, а до 56 остается 2). Здесь мы проверяем условие равенства остатка 0, так как четные числа – те, что делятся на 2 без остатка, а кратные 3 – те, что делятся на 3 без остатка).

```python
# оператор ИСКЛЮЧАЮЩЕЕ ИЛИ
# верно ровно одно условие: либо только первое, либо только второе
# выбираем четные элементы или кратные 3, исключая те, что делятся и на 2 и на 3 сразу,
# по сравнению с предыдущим результатом выкинули 12, 24, 30

chosen_two = sample[(sample % 2 == 0) ^ (sample % 3 == 0)]
print(chosen_two)
```

### Задача 2. Экспериментальная

В массиве `group` зафиксировано, к какой группе относится участник эксперимента, посвященного эффективности медицинского препарата. Группы A и B являются тестовыми – участники принимают препарат в разных дозировках, а группа C является контрольной – участники принимают плацебо.

```python
group = np.array(["A", "A", "A", "B", "C", "A", "B", "B", "B", "B", 
                  "C", "A", "A", "C", "C", "C", "A", "B", "B", "A", 
                  "A", "A", "B", "B", "B", "C", "A", "A", "B", "C",
                  "B", "C", "A", "B", "A", "C", "B", "B", "C", "B", 
                  "B", "C", "A", "C", "C", "A", "A", "C", "B", "A"])
```

Выберите из массива `group` представителей групп A и B и сохраните их в массив `test`, остальных сохраните в массив `control`.

```python
test = group[(group == "A") | (group == "B")]
print(test)
```

```python
control = group[group == "C"]
print(control)
```

```python
# как вариант, раз кроме A, B, C значений нет,
# можно test получить просто отрицанием C

test = group[(group != "C")]
print(test)
```
