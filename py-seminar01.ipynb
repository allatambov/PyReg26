### Немного теории: функции, методы и атрибуты

```python
import numpy as np
```

```python
grades = [10, 9, 6, 5, 8]
rating = np.array([4.5, 9.8, 6.5, 8.2, 9.3, 4.8, 5.2, 6.3])
lullaby = "Спят усталые питоны, змеи спят... И усталые студенты спать хотят"

print(grades)
print(rating)
print(lullaby)
```

Функции

```python
print(len(grades))
print(len(rating))
print(len(lullaby))
```

```python
print(sorted(grades))
print(sorted(rating))
print(sorted(lullaby))
```

```python
print(sum(grades))
print(sum(rating))
```

Методы

```python
# на строках

print(lullaby.upper())
print(lullaby.lower())
print(lullaby.split("..."))
```

```python
# на массивах

print(rating.mean())
print(rating.var())
print(rating.reshape((2, 4)))
```

```python
# атрибуты

print(rating.dtype)
print(rating.size)
print(rating.ndim)
```
