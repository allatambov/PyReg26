```python
# удаление строк с пропусками
# удаляются строки, где есть пропуски в столбцу Eye

beasts = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyReg26/refs/heads/main/fbeasts.csv")
beasts = beasts.dropna(subset = ["Eye"])
print(beasts.shape)
```

### Добавление новых столбцов

```python
# домножаем столбец на число и добавляем в таблицу
# с названием Size_cm

beasts["Size_cm"] = beasts["Size"] * 2.54
beasts
```

```python
# проверяем условие, получаем столбец из True/False,
# превращаем их в 1 и 0, превращая столбец в целочисленный int

beasts["Danger"] = (beasts["Classification"] == "XXXXX").astype(int)
beasts
```

```python
# считаем количество символов в каждой ячейке
# текстового столбца Classification

beasts["ClassNum"] = beasts["Classification"].str.len()
beasts
```

```python
# в str много разных функций,
# делаем все буквы маленькими - lower()

beasts["Classification"].str.lower()
```

```python
# в str много разных функций,
# проверяем вхождение подстроки Russia в ячейки с текстом

beasts["Native"].str.contains("Russia")
```

```python
# подставляем True/False выше
# в фильтр и отбираем строки с Russia в Native

beasts[beasts["Native"].str.contains("Russia")]
```
