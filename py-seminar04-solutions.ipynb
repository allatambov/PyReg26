Выберем отдельный столбец и опишем его:

```python
df["Size"].describe()
```

> Для выбора столбца его название указывается в квадратных скобках в кавычках. Такой синтаксис объясняется тем, что в Python объект типа `pandas DataFrame` изнутри выглядит как словарь (базовый тип `dictionary` или просто `dict`). Словарь – набор пар «ключ-значение», и выбор значений производится по ключу. Так, таблица в `df` в «сыром» виде похожа на словарь, где ключами выступают названия столбцов, а значениями – столбцы с данными:

    {"Name" : np.array(["Acromantula", "Basilisk", "Chimaera", ...]), 
    "Class" : np.array(["Acromantula", "Basilisk", "Chimaera", ...]),
    "Classification" : np.array(["XXXXX", "XXXXX", "XXXXX", ...])}

Если написать `df[Size]`, мы получим ошибку `NameError`, так как без кавычек Python будет воспринимать `Size` как название переменной, а мы такой переменной не создавали. Если написать просто `"Size"` без `df`, для Python это просто текст со словом "Size". 

**Дополнение.** Каждый столбец датафрейма – это объект типа `pandas Series`, то есть последовательность Pandas.

```python
df["Dragon"]
```

Этот объект – «усовершенствованный» массив NumPy. Он состоит из двух частей – индексы элементов (0, 1, ..., 80) и сами элементы. Индексы элементов хранятся в атрибуте `.index`, а значения – в атрибуте `.values`:

```python
# .index – отдельный тип RangeIndex
# .values – массив NumPy

print(df["Dragon"].index)
print(df["Dragon"].values)
```

Для фильтрации строк или добавления новых столбцов забирать элементы отдельно через `.values` не потребуется, однако понимание внутреннего устройства столбца датафрейма позволит понять, почему в некоторых методах (например, для сортировки) нужно уточнять, с чем мы работаем, с индексами или самими значениями.

### Задача 1

Выберите строки, соответствующие драконам, и сохраните эти строки в датафрейм `dragons`.

```python
dragons = df[df["Dragon"] == 1]
dragons
```

### Задача 2

Выберите строки, соответствующие существам из Греции или Румынии (*Greece* или *Romania* в `Native`). Вычислите их количество.

### Задача 3

Выберите строки, соответствующие существам размером более 500 дюймов.

```python
df3 = df[df["Size"] > 500]
df3
```

### Задача 4

Выберите строки, соответствующие существам размером от 100 до 200 дюймов включительно.

### Задача 6

Выберите строки, соответствующие существам с классом Министерства Магии не ниже 4 и имеющим желтые глаза (`Yellow`). Сохраните эти строки в датафрейм `dang_yellow`.

```python
# внимание на дополнительные круглые скобки вокруг
# условий с ИЛИ – для порядка действий

dang_yellow = df[((df["Classification"] == "XXXX") | 
                  (df["Classification"] == "XXXXX")) & 
    (df["Eye"] == "Yellow")]
dang_yellow
```

```python
# альтернативное решение
# метод .isin() проверяет, входит ли элемент в перечень

dang_yellow = df[df["Classification"].isin(["XXXX", "XXXXX"])
                   & (df["Eye"] == "Yellow")]
dang_yellow
```

**Дополнение.** В этих вариантах решения мы пока не рассматриваем подсчет количества символов в ячейках столбца `Classification` (об этом поговорим в следующий раз), а пользуемся тем, что не ниже 4 – это `XXXX` или `XXXXX`. Чтобы убедиться, что других значений нет, можем запросить таблицу частот через метод `.value_counts()`:

```python
df["Classification"].value_counts()
```

По умолчанию используется сортировка записей по убыванию частот – самое частое значение в столбце `XXX`, оно встречается 27 раз, далее идет `XXXX` – встречается 20 раз, и так далее. Если мы хотим отсортировать категории по алфавиту (здесь от `X` до `XXXXX`), потребуется метод `.sort_index()`:

```python
df["Classification"].value_counts().sort_index()
```

Почему именно `.sort_index()`, а не просто `.sort()`? Результат `.value_counts()` – отдельный столбец, последовательность `pandas Series`, состоящая из двух частей, из индексов и значений. Поэтому при сортировке надо уточнять, на основании чего упорядочивать элементы.

```python
# разберем на части результат .value_counts()
# ровно то, что было перед задачей 1
# круг замкнулся :)

tab = df["Classification"].value_counts()
print(tab.index)
print(tab.values)
```

```python
# сортировка по индексам vs сортировка по значениям
# по умолчанию по алфавиту или по возрастанию 

print(tab.sort_index())
print(tab.sort_values())
```
