* `BorrName`: название компании-заемщика;
* `BorrStreet`, `BorrCity`, `BorrState`: адрес, город, штат компании-заемщика;

* `LoanOk`: количество одобренных или выплаченных кредитов за предыдущие 10 лет (2010-2019 годы).
